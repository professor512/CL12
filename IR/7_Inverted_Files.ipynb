```python
# Assignment No. 07
# Document Retrieval using Inverted Index

# Define documents
document1 = "The quick brown fox jumped over the lazy dog."
document2 = "The lazy dog slept in the sun."
document3 = "The quick fox is running in the forest."

# Step 1: Tokenize documents
tokens1 = document1.lower().replace(".", "").split()
tokens2 = document2.lower().replace(".", "").split()
tokens3 = document3.lower().replace(".", "").split()

# Step 2: Build inverted index
inverted_index = {}

documents = {
    "Document 1": tokens1,
    "Document 2": tokens2,
    "Document 3": tokens3
}

for doc_name, tokens in documents.items():
    for word in tokens:
        if word not in inverted_index:
            inverted_index[word] = []
        if doc_name not in inverted_index[word]:
            inverted_index[word].append(doc_name)

# Step 3: Display inverted index
print("INVERTED INDEX")
print("----------------")

for word in sorted(inverted_index):
    print(word, "->", ", ".join(inverted_index[word]))

# Step 4: Document retrieval
query = input("\nEnter a word to search: ").lower()

if query in inverted_index:
    print("\nDocuments containing", query, ":")
    for doc in inverted_index[query]:
        print(doc)
else:
    print("\nNo document found.")
```
